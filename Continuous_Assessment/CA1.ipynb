**Conceptual Details**
A Simple Reflex Agent is an intelligent agent that decides what to do based only on the current condition of the environment. It does not remember previous conditions or try to predict what will happen next.

In this practical, the environment is a theatre/auditorium with two locations, A and B. Each location can randomly have one of four conditions:


*   Clean - no cleaning is required.
*   Dusty - the agent performs dust cleaning.
*   Popcorn Waste - the agent activates waste collection.
*   Drink Spill - the agent performs spill cleaning.

The vacuum starts at a randomly selected location and checks its current condition. Based on that condition, it immediately performs the appropriate action. After cleaning the first location, it moves to the other location and performs the same process.

The performance of the agent is measured using a score. Cleaning gives positive points, while movement between locations reduces the score slightly.

The basic working can be represented as:

Current Condition → Appropriate Action → Location Becomes Clean

For example:

Popcorn Waste → Waste Collection → Clean

This makes the system a simple reflex agent because its action depends only on what it currently senses.

**Algorithm**
Step 1: Start the program.

Step 2: Create the theatre environment with two locations, A and B.

Step 3: Assign a random condition to each location from the following:

Clean
Dusty
Popcorn Waste
Drink Spill

Step 4: Randomly place the vacuum cleaner at either Location A or Location B.

Step 5: Check the current condition of the location.

Step 6: If the location is Clean, take no cleaning action.

Step 7: If the location is Dusty, activate the dust cleaning action and make the location clean.

Step 8: If the location contains Popcorn Waste, activate waste collection and make the location clean.

Step 9: If the location has a Drink Spill, activate spill cleaning and make the location clean.

Step 10: Update the performance score according to the action performed.

Step 11: Move the vacuum to the other location and reduce the score for movement.

Step 12: Check the condition of the second location and perform the appropriate cleaning action.

Step 13: Display the final condition of both theatre locations.

Step 14: Display the final performance score.

Step 15: Stop the program.

**Conclusion:**

The Theatre Vacuum Agent successfully cleans both theatre locations based on their floor conditions.
It uses a simple reflex approach by taking actions according to the current condition of each location.
Different cleaning modes are activated for dust, popcorn waste, and drink spills.
The agent updates each cleaned location to Clean after performing the required action.
A performance score is calculated based on cleaning performed and movement between locations.
Thus, the system demonstrates how a simple reflex agent can efficiently maintain a clean theatre environment.

```python
import random

class Environment:

    def __init__(self):

        # Theatre floor conditions
        # 0 = Clean
        # 1 = Dusty
        # 2 = Popcorn Waste
        # 3 = Drink Spill

        self.locationCondition = {
            'A': random.randint(0, 3),
            'B': random.randint(0, 3)
        }


class TheatreVacuumAgent:

    def __init__(self, environment):

        print("Initial Theatre Environment:")
        print(environment.locationCondition)

        # Performance measurement
        score = 0

        # Randomly place vacuum at A or B
        currentLocation = random.choice(['A', 'B'])

        print("\nVacuum starts at Location", currentLocation)

        # Visit both theatre sections
        locations = [currentLocation]

        if currentLocation == 'A':
            locations.append('B')
        else:
            locations.append('A')

        # Clean both locations
        for location in locations:

            condition = environment.locationCondition[location]

            print("\nChecking Location", location)

            if condition == 0:
                print("Location", location, "is Clean.")
                print("No cleaning required.")

            elif condition == 1:
                print("Location", location, "is Dusty.")
                print("Activating Dust Cleaning Mode...")

                environment.locationCondition[location] = 0
                score += 1

                print("Dust has been removed.")

            elif condition == 2:
                print("Location", location, "has Popcorn Waste.")
                print("Activating Waste Collection Mode...")

                environment.locationCondition[location] = 0
                score += 2

                print("Popcorn waste has been collected.")

            elif condition == 3:
                print("Location", location, "has a Drink Spill.")
                print("Activating Spill Cleaning Mode...")

                environment.locationCondition[location] = 0
                score += 2

                print("Drink spill has been cleaned.")

            # Move to next location
            if location != locations[-1]:
                print("Moving to Location", locations[1])

                score -= 1

        print("\nFinal Theatre Environment:")
        print(environment.locationCondition)

        print("\nPerformance Measurement:", score)


# Create the theatre environment
theEnvironment = Environment()

# Create the simple reflex vacuum agent
theVacuum = TheatreVacuumAgent(theEnvironment)
```
